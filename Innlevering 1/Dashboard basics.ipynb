```python
import pandas as pd

df = pd.read_csv("../data/reservoirs.csv")

df.head()
```

```python
print(df.columns)
df.info()
```

```python
df = df.rename(columns={
    "dato_Id": "date",
    "omrType": "area_type",
    "omrnr": "area_number",
    "iso_aar": "year",
    "iso_uke": "week",
    "fyllingsgrad": "filling_ratio",
    "kapasitet_TWh": "capacity_TWh",
    "fylling_TWh": "stored_energy_TWh",
    "neste_Publiseringsdato": "next_publication_date",
    "fyllingsgrad_forrige_uke": "previous_week_filling_ratio",
    "endring_fyllingsgrad": "filling_ratio_change"
})
```

```python
print(df.columns)
```

```python
df.head()
```

```python
display(df.head())

df.info()

display(df.describe())
```

```python
import matplotlib.pyplot as plt

# Gjør om dato til datetime
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values(by='date')

# Numeriske kolonner
numeric_columns = [
    "area_number",
    "year",
    "week",
    "filling_ratio",
    "capacity_TWh",
    "stored_energy_TWh",
    "previous_week_filling_ratio",
    "filling_ratio_change"
]

# Ett plot per kolonne
for col in numeric_columns:
    
    plt.figure(figsize=(10, 4))
    
    plt.plot(df["date"], df[col])
    
    plt.title(f"{col} over time")
    plt.xlabel("Date")
    plt.ylabel(col)
    
    plt.grid(True)
    plt.tight_layout()
    
    plt.show()
```

```python
import matplotlib.pyplot as plt

# Kun Norge
df_no = df[df["area_type"] == "NO"].copy()

# Sorter etter dato
df_no = df_no.sort_values("date")

plt.figure(figsize=(12, 6))

plt.plot(
    df_no["date"],
    df_no["filling_ratio"],
    label="filling_ratio"
)

plt.plot(
    df_no["date"],
    df_no["previous_week_filling_ratio"],
    label="previous_week_filling_ratio"
)

plt.title("Reservoir Filling Ratio for Norway")
plt.xlabel("Date")
plt.ylabel("Ratio")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df_no["date"],
    df_no["capacity_TWh"],
    label="capacity_TWh"
)

plt.plot(
    df_no["date"],
    df_no["stored_energy_TWh"],
    label="stored_energy_TWh"
)

plt.title("Reservoir Energy Storage for Norway")
plt.xlabel("Date")
plt.ylabel("Energy (TWh)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df_no["date"],
    df_no["filling_ratio_change"],
    label="filling_ratio_change"
)

plt.title("Weekly Change in Reservoir Filling Ratio")
plt.xlabel("Date")
plt.ylabel("Change in Filling Ratio")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
